```python
from sentence_transformers import SentenceTransformer
import pandas as pd
```

```python
model_bge_m3 = SentenceTransformer("BAAI/bge-m3")
```

```python
sentences = [
    "Ich suche eine Stelle im Bereich Datenanalyse.",
    "Wir suchen einen Data Analyst mit Python- und SQL-Kenntnissen.",
    "Wir suchen einen Koch für unser Restaurant."
]

embeddings = model_bge_m3.encode(sentences)

print(embeddings.shape)
```
